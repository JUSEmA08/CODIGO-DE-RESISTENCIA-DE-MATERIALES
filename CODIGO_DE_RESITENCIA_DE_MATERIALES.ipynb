```python
#CALCULADORA DE ESFUERZOS - PARA (COMPRESION,TRACCION Y CORTANTE)
#UNIDADES:
#fuerza: N
#longitud: mm
#esfuerzos:MPa
#E: GPa
#momentos o fuerzas de par: Nm
#cargas distribuidas W : N/mm

from math import pi, sqrt

def numero(texto):
  while True:
    try:
      valor = float(input(texto))
      return valor
    except ValueError:
      print("porfavor, ingresa un número válido.")

#COMPRESION Y TRACCION

def axial():
    print("\n--- TRACCIÓN / COMPRESIÓN ---")
    P = numero("Fuerza P en N (positiva = tracción, negativa = compresión): ")
    L = numero("Longitud L en mm: ")
    A = numero("Área de la sección A en mm²: ")
    E = numero("Módulo de elasticidad E en GPa (acero = 200): ")

    E = E*1000
    ESFUERZOS = P/A
    DEFORMACION = ESFUERZOS/E
    CAMBIO = DEFORMACION*L

    print("\nResultados:")
    print(f"Esfuerzo es: {ESFUERZOS:.4f} MPa")
    print(f"Deformación unitaria es: {DEFORMACION:.6f}")
    print(f"Cambio de longitud es: {CAMBIO:.4f} mm")
    if P >= 0:
      print("Es una tracción.")
    else:
      print("Es una compresión.")
      Sy = numero("resistencia del material Sy en Mpa,(0 si no desea factor de seguridad: ")
      if Sy > 0:
        print("factor de seguridad:", round(abs(Sy/ESFUERZOS), 2))

# FUERZA CORTANTE
def cortante():
    print("\n--- FUERZA CORTANTE ---")
    print("\n1) Perno o Pasador")
    print("\n2) Cortante en vigas")

    OP = input("Elija las opciones 1 o 2: ")
    if OP == "1":
        print("\n--- FUERZA CORTANTE ---")
        F = numero("Fuerza F en N: ")
        D = numero("Diametro D del perno en mm: ")
        n = numero("cantidad de pernos: ")
        planos = numero("Planos de corte (1: simple, 2:doble): ")

        area = (pi*D**2)/4
        tau = F/(n*planos*area)
        print("\nResultados:")
        print(f"Esfuerzo es: {tau:.4f} MPa")

    elif OP == "2":
        print("\n--- FUERZA CORTANTE ---")
        V = numero ("fuerza cortante en N: ")
        print("seccion: 1) rectangular, 2) circular")
        s = input("selecciona la seccion 1 o 2 : ")
        if s == "1":
          b = numero("base b en mm: ")
          h = numero("altura h en mm: ")
          area = b*h
          tau_max = 1.5*V/area
          print("\nResultados:")
          print(f"Esfuerzo máximo (viga rectangular) es: {tau_max:.4f} MPa")
        elif s == "2":
          d = numero("diametro d en mm: ")
          area = pi*(d/2)**2
          tau_max = (4*V)/(3*area)
          print("\nResultados:")
          print(f"Esfuerzo máximo (viga circular) es: {tau_max:.4f} MPa")
        else:
          print("opcion invalida")

  #MENÚ DE ENTRADA

print("--- CALCULADORA DE ESFUERZOS ---")
print("\n================================")
print("\n1) TRACCIÓN / COMPRESIÓN")
print("\n================================")
print("\n2) FUERZA CORTANTE")
print("\n================================")
OP = input("Elija las opciones 1 o 2: ")
if OP == "1":
    axial()
elif OP == "2":
    cortante()
else:
    print("opcion invalida")
```
